```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import glob
import os

BASE_DIR = "/content/drive/MyDrive/AI351ProjectTest/05-FinalCompilation/"
os.makedirs(BASE_DIR, exist_ok=True)

DOWN_PATH = "/content/drive/MyDrive/AI351ProjectTest/04-DataAugmentation/downsampled.csv"
AUG_PATH_PATTERN = "/content/drive/MyDrive/AI351ProjectTest/04-DataAugmentation/Augmentation*.csv"
```

```python
df_down = pd.read_csv(DOWN_PATH)
df_down["Flag"] = 0

print("Downsampled loaded:", df_down.shape)
df_down.head()
```

```python
aug_files = sorted(glob.glob(AUG_PATH_PATTERN))

compiled_augmented = []

for file in aug_files:
    # Extract number (e.g., "Augmentation03.csv" → 3)
    filename = os.path.basename(file)
    num = int(filename.replace("Augmentation", "").replace(".csv", ""))

    df_aug = pd.read_csv(file)
    df_aug["Flag"] = num

    compiled_augmented.append(df_aug)
    print(f"Loaded {filename} → Flag = {num}, Shape = {df_aug.shape}")

print("\nTotal augmentation files loaded:", len(compiled_augmented))
```

```python
df_final = pd.concat([df_down] + compiled_augmented, ignore_index=True)

print("FINAL dataset shape:", df_final.shape)
df_final.head()
```

```python
OUTPUT_PATH = BASE_DIR + "FinalCompiledDataset.csv"
df_final.to_csv(OUTPUT_PATH, index=False)

print("Saved final merged dataset to:")
print(OUTPUT_PATH)
```
